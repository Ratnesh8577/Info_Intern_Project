# 🛫Flight Counts by Route and Month (Top 10 Routes)

```python
# Pivot table: rows=Route, columns=Month, values=FlightCount
route_month_counts = df.pivot_table(
    index='Route',
    columns='Month',
    values='FlightNum',
    aggfunc='count',
    fill_value=0
)

# Select top 10 busiest routes (ensure top_routes exists)
top_routes_list = top_routes['Route'].tolist()

# Filter only routes that actually exist in pivot table
existing_routes = [r for r in top_routes_list if r in route_month_counts.index]

if existing_routes:
    route_month_counts = route_month_counts.loc[existing_routes]
    
    # 🟩 Show the actual data values
    print("✅ Route vs Month Flight Counts:")
    display(route_month_counts)

    # Plot heatmap
    plt.figure(figsize=(14, 6))
    sns.heatmap(route_month_counts, annot=True, fmt='d', cmap='YlGnBu')
    plt.title('Flight Counts by Route and Month (Top 10 Routes)')
    plt.xlabel('Month')
    plt.ylabel('Route')
    plt.tight_layout()
    display(plt.gcf())
    plt.close()
else:
    print("⚠️ No matching routes found between top_routes and route_month_counts.")
```

```python
# Compute average delay for top routes
route_delays = df.groupby('Route')['TotalDelay'].mean().loc[top_routes['Route']]

# Use a cooler gradient color palette
colors = sns.color_palette("crest", n_colors=len(route_delays))

plt.figure(figsize=(12, 7))
bars = sns.barplot(
    x=route_delays.values,
    y=route_delays.index,
    palette=colors,
    edgecolor='black',
    linewidth=1.2
)

# Add annotations on bars
for bar in bars.patches:
    plt.text(
        bar.get_width() + 1,                # position text slightly after bar end
        bar.get_y() + bar.get_height() / 2, # vertically centered
        f"{bar.get_width():.1f}",           # format: one decimal
        va='center',
        fontsize=10,
        color='black'
    )

# Style adjustments
plt.title("🌈 Average Total Delay for Top 10 Routes", fontsize=16, fontweight='bold', pad=15)
plt.xlabel("Average Delay (minutes)", fontsize=12)
plt.ylabel("Route", fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()
display(plt.gcf())
plt.close()

     
```

# 3)  🌍Maps showing busiest airports and average delays

```python
# Get unique origin airports
unique_airports = df["Org_Airport"].unique()

# Assign random coordinates (latitude between 25–49, longitude between -125–67)
coords = {a: (np.random.uniform(25, 49), np.random.uniform(-125, -67)) for a in unique_airports}

# Map coordinates back to DataFrame
df["Latitude"] = df["Org_Airport"].map(lambda x: coords[x][0])
df["Longitude"] = df["Org_Airport"].map(lambda x: coords[x][1])

# Display sample to verify
df[["Org_Airport", "Latitude", "Longitude"]].head(10)
```

```python
import plotly.express as px
import pandas as pd

# Ensure required columns exist
required_cols = ['Org_Airport', 'Latitude', 'Longitude', 'ArrDelay']
if all(col in df.columns for col in required_cols):

    # 1️⃣ Aggregate airport-level metrics
    airport_stats = (
        df.groupby(['Org_Airport', 'Latitude', 'Longitude'])
          .agg(
              Flight_Count=('Org_Airport', 'size'),
              Avg_ArrDelay=('ArrDelay', 'mean')
          )
          .reset_index()
    )

    # 2️⃣ Map of busiest airports (by Flight_Count)
    fig1 = px.scatter_geo(
        airport_stats,
        lat='Latitude',
        lon='Longitude',
        size='Flight_Count',
        color='Flight_Count',
        hover_name='Org_Airport',
        projection='natural earth',
        title='✈️ Busiest Airports by Flight Count',
        color_continuous_scale='Plasma'
    )
    fig1.show()

    # 3️⃣ Map of average arrival delay (by airport)
    fig2 = px.scatter_geo(
        airport_stats,
        lat='Latitude',
        lon='Longitude',
        size='Avg_ArrDelay',
        color='Avg_ArrDelay',
        hover_name='Org_Airport',
        projection='natural earth',
        title='⏰ Average Arrival Delay by Airport (minutes)',
        color_continuous_scale='RdYlBu_r'
    )
    fig2.show()

else:
    print("⚠️ Required columns not found: Please include Org_Airport, Latitude, Longitude, and ArrDelay.")
```

# Average Cancellation Rate: Winter vs Non-Winter

```python
# Define winter months
winter_months = ['December', 'January', 'February']

# Ensure 'Date' is in datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Extract month names from Date
df['Month'] = df['Date'].dt.month_name()

# Calculate average cancellation rate per month
cancel_rate = df.groupby('Month')['Cancelled'].mean().reset_index()

# Label each month as Winter or Non-Winter
cancel_rate['Season'] = cancel_rate['Month'].apply(
    lambda x: 'Winter' if x in winter_months else 'Non-Winter'
)

# Compute average cancellation rate by season
seasonal_summary = cancel_rate.groupby('Season')['Cancelled'].mean()
print(seasonal_summary)

# Plot comparison
seasonal_summary.plot(
    kind='bar',
    title='Average Cancellation Rate: Winter vs Non-Winter',
    ylabel='Cancellation Rate',
    color=['#4C72B0', '#55A868'],
    figsize=(8, 5)
)

plt.show()
```

# ✈️Airline Cancellation Rate:🌨️Winter vs 🔥Non-Winter

```python
# Define winter months
winter_months = ['December', 'January', 'February']

# Ensure 'Date' is datetime
df['Date'] = pd.to_datetime(df['Date'])

# Extract month and define season
df['Month'] = df['Date'].dt.month_name()
df['Season'] = df['Month'].apply(lambda x: 'Winter' if x in winter_months else 'Non-Winter')

# Calculate cancellation rate by Airline and Season
airline_season_cancel = (
    df.groupby(['Airline', 'Season'])['Cancelled']
      .mean()
      .reset_index()
)

# Pivot table for plotting
pivot_airline_season = airline_season_cancel.pivot(
    index='Airline',
    columns='Season',
    values='Cancelled'
)

# Plot comparison
pivot_airline_season.plot(
    kind='bar',
    figsize=(12, 6),
    title='Airline Cancellation Rate: Winter vs Non-Winter',
    ylabel='Cancellation Rate'
)

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
